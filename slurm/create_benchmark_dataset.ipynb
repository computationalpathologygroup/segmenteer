```python
from pathlib import Path

import pandas as pd
```

```python
IMAGE_ROOT = Path("/mnt/pa_cpgarchive/projects/tissue-segmentation/data/benchmark/slides/")
ANNOTATION_ROOT = Path("/mnt/pa_cpgarchive/projects/tissue-segmentation/data/benchmark/geojsons/")

images = sorted(IMAGE_ROOT.glob("*.tif"))
annotations = sorted(ANNOTATION_ROOT.glob("*.geojson"))

for image, annotation in zip(images, annotations):
    assert image.stem == annotation.stem, f"Image and annotation names do not match: {image.stem} != {annotation.stem}"

dataset_df = pd.DataFrame({
    "path": [str(image) for image in images],
})
dataset_df.to_csv("datasets/benchmark_dataset.csv", index=False)

annotation_mappings_df = pd.DataFrame({
    "slide_id": [image.stem for image in images],
    "ground_truth_annotation": [str(annotation) for annotation in annotations],
})
annotation_mappings_df.to_csv("annotation-mappings/benchmark_annotations.csv", index=False)
```

```python
CPG_DATATEAM_ROOT = Path("/mnt/pa_cpgarchive/projects/tissue-segmentation/data/data_team/slides")
cpg_datateam_images = sorted(CPG_DATATEAM_ROOT.glob("*.tif"))
cpg_datateam_dataset_df = pd.DataFrame({
    "path": [str(image) for image in cpg_datateam_images],
})
cpg_datateam_dataset_df.to_csv("datasets/cpg_datateam_dataset.csv", index=False)


CPG_CROWDSOURCE_ROOT = Path("/mnt/pa_cpgarchive/projects/tissue-segmentation/data/crowdsource/slides")
cpg_crowdsource_images = sorted(CPG_CROWDSOURCE_ROOT.glob("*.tif"))
cpg_crowdsource_dataset_df = pd.DataFrame({
    "path": [str(image) for image in cpg_crowdsource_images],
})
cpg_crowdsource_dataset_df.to_csv("datasets/cpg_crowdsource_dataset.csv", index=False)
```
